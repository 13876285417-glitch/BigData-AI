## 4. 新闻场景小练习：统计一篇短稿的字数

纸媒时代的编辑要手动「数字数」来控制版面，现在交给程序一秒完成。

```python
news_text = '''国家统计局今日发布数据，8月份全国居民消费价格指数同比上涨0.6%，
涨幅比上月扩大0.1个百分点。其中，食品价格上涨2.8%，非食品价格上涨0.2%。'''

# len() 能数出字符串里有多少个字符
total_chars = len(news_text)

# 去掉换行符和空格，得到「净字数」
clean_text = news_text.replace('\n', '').replace(' ', '')
clean_chars = len(clean_text)

print('原始字符数（含换行与空格）：', total_chars)
print('净字数：', clean_chars)
```
